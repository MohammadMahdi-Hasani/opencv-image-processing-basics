```python
## Part 3 ## 

## Using matplotlib ##

img = cv2.imread('Camman256B.bmp',0)
plt.imshow(img, cmap = 'gray', interpolation = 'bicubic')
plt.xticks([]), plt.yticks([])  # to hide tick values on X and Y axis
plt.show()

## Making borders for image (padding) ##

Color=[255,0,0]
 
 
replicate=cv2.copyMakeBorder(img,13,15,17,19,cv2.BORDER_REPLICATE)
 
reflect=cv2.copyMakeBorder(img,13,15,17,19,cv2.BORDER_REFLECT)
 
reflect101=cv2.copyMakeBorder(img,3,5,7,9,cv2.BORDER_REFLECT_101 )
 
wrap=cv2.copyMakeBorder(img,13,15,17,19,cv2.BORDER_WRAP )
 
constant=cv2.copyMakeBorder(img,13,15,17,19,cv2.BORDER_CONSTANT,value=Color)
 
 
plt.subplot(231),plt.imshow(img,'gray'),plt.title('ORIGINAL')
 
plt.subplot(232),plt.imshow(replicate,'gray'),plt.title('REPLICATE')
 
plt.subplot(233),plt.imshow(reflect,'gray'),plt.title('REFLECT')
 
plt.subplot(234),plt.imshow(reflect101,'gray'),plt.title('REFLECT_101')
 
plt.subplot(235),plt.imshow(wrap,'gray'),plt.title('WRAP')
 
plt.subplot(236),plt.imshow(constant,'gray'),plt.title('CONSTANT')
 
 
plt.show()
```
